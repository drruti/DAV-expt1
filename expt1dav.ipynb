```python

import numpy as np
import pandas as pd

a = np.array([10, 20, 30, 40, 50])

print("Array:", a)
print("Mean:", a.mean())
print("Reshaped 2D:\n", a.reshape(5, 1))
print("Sliced:", a[1:4])

data = {
    "Name": ["Drruti", "Pratheek", "Senthil", "Chitra"],
    "Marks": [93, 95, 94, 96],
    "Branch": ["CSE", "CSE", "CSE", "CSE"]
}

df = pd.DataFrame(data)

print(df)

print("\nFirst 2 rows:\n", df.iloc[0:2])

print("\nMarks column:\n", df["Marks"])

print("\nRows where marks > 70:\n", df[df["Marks"] > 70])
```
